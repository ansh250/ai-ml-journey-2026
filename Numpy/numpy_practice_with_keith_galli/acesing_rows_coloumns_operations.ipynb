Be Carefull When Copying Arrays

```python
a = np.array([1,2,3])
#copy the array a in b
b = a
print(b)
```

```python
#change the b array value
b[0] = 150
b
```

```python
# now again print a
a
```

```python
# we can see we are changing in b array but it also changes the value of a array
# so for stoping this 
b=a.copy()
b[0] = 500
b
```

```python
a # .copy() function stops the change ment in original array
```

##### Reorganizing Arrays

```python
before = np.array([[1,2,3,4],[5,6,7,8]])
print(before)
```

```python
print(before.shape)
```

```python
after = before.reshape(4,2)
print(after)
```

```python
# Vertically stacking vectors
v1 = np.array([1,2,3])
v2 = np.array([4,5,6])
# np.vstack([v1,v2])
np.vstack([v1,v2,v2,v1])
```

```python
#Horizontal stack
h1 = np.ones((2,4))
h2 = np.zeros((2,2))
np.hstack((h1,h2))
```
